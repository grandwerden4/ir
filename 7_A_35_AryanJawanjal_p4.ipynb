```python

documents = {
    "D1": 'Machine Learing improves search engines',
    'D2': 'Deep Learning and machine intelligence',
    'D3': 'Search Systems use dictionary indexing',
    'D4': 'Prefix matching is useful in retrieval',
    'D5': 'Wildcard search support tolerant retrieval',
    'D6': 'Moon Mould Mouse Mood  Accept Affect Principlr Principal'
}

# convert to lowercase
for doc in documents:
    documents[doc] = documents[doc].lower()

# create dictionary
vocabulary = set()

for text in documents.values():
    words = text.split()
    vocabulary.update(words)

vocabulary = sorted(vocabulary)
print("Dictionary")
print(vocabulary)

while True:
    print("\n1. Prefix Search")
    print("2. Wildcard Search")
    print("3. Exit")
    choice = input("Enter your choice: ")
    if choice == '1':
        prefix = input("Enter the prefix to search: ")
        result = [word for word in vocabulary if word.startswith(prefix.lower())]
        print("Prefix search results:", result)
    elif choice == '2':
        wildcard = input("Enter the wildcard to search: ")
        import fnmatch
        result = [word for word in vocabulary if fnmatch.fnmatch(word, wildcard.lower())]
        print("Matched words:", result)
    elif choice == '3':
        break
    else:
        print("Invalid choice")
```

```python
import fnmatch
import os

# Path to an existing text file in Colab
file_path = '/content/sample_data/README.md'

if os.path.exists(file_path):
    with open(file_path, 'r') as f:
        text_content = f.read()

    # Build vocabulary from the existing document
    # Removing some common punctuation and splitting into words
    words = text_content.lower().replace('*', '').replace('`', '').replace('[', '').replace(']', '').replace('(', '').replace(')', '').split()
    vocabulary = sorted(list(set(words)))

    print(f"Successfully loaded: {file_path}")
    print(f"Vocabulary size: {len(vocabulary)} unique words")
    print("Preview:", vocabulary[:10])

    # Search Interface
    while True:
        print("\n1. Prefix Search")
        print("2. Wildcard Search")
        print("3. Exit")
        choice = input("Enter choice (1-3): ")

        if choice == '1':
            prefix = input("Enter prefix: ").lower()
            results = [w for w in vocabulary if w.startswith(prefix)]
            print("Results:", results)
        elif choice == '2':
            pattern = input("Enter wildcard pattern (e.g. *data*): ").lower()
            results = [w for w in vocabulary if fnmatch.fnmatch(w, pattern)]
            print("Results:", results)
        elif choice == '3':
            print("Exiting.")
            break
        else:
            print("Invalid choice.")
else:
    print(f"File {file_path} not found.")d
```

```python
# collection of 10 documents

documents = {
    1: "Artificial Intelligence is used in modern technology.",
    2: "Python is a popular programming language.",
    3: "Machine Learning is a branch of Artificial Intelligence.",
    4: "Data Science uses Python for data analysis.",
    5: "Cloud Computing provides scalable computing resources.",
    6: "Python is widely used in Machine Learning.",
    7: "Artificial Intelligence helps in data science applications.",
    8: "Cloud Computing supports Artificial Intelligence systems.",
    9: "Data Science and Machine Learning are important fields.",
    10: "Python programming is useful for Artificial Intelligence."
}

# dictionary of unique words

import re

dictionary = set()

for docid, text in documents.items():
    text = text.lower()
    words = re.findall(r'\w+', text)
    dictionary.update(words)

dictionary = sorted(dictionary)

print("Dictionary:")
print(dictionary)
```

```python

# inverted index

import fnmatch

inverted_index = {}

for term in dictionary:
    inverted_index[term] = []
    for docid, text in documents.items():
        words = re.findall(r'\w+', text.lower())
        if term in words:
            inverted_index[term].append(docid)

print("\nInverted index ")
print("-" * 40)

for term in inverted_index:
    print(f"{term:15} -> {inverted_index[term]}")



while True:
    print("\n")
    print("=" * 50)
    print("1. Exact search")
    print("2. Prefix search")
    print("3. Wildcard search")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        query = input("Enter search term: ").lower()

        if query in inverted_index:
            print("\nDocuments Found:")
            for doc in inverted_index[query]:
                print(doc, ":", documents[doc])
        else:
            print("No documents found")

    elif choice == "2":
        prefix = input("Enter prefix: ").lower()
        matched_terms = []

        for term in dictionary:
            if term.startswith(prefix):
                matched_terms.append(term)

        if matched_terms:
            print("\nMatching terms:", matched_terms)
            docs = set()

            for term in matched_terms:
                docs.update(inverted_index[term])

            print("\nDocuments Found:")
            for doc in docs:
                print(doc, ":", documents[doc])
        else:
            print("No matching terms found")

    elif choice == "3":
        pattern = input("Enter wildcard pattern (* and ? supported): ").lower()

        matched_terms = []

        for term in dictionary:
            if fnmatch.fnmatch(term, pattern):
                matched_terms.append(term)

        if matched_terms:
            print("\nMatching terms:", matched_terms)
            docs = set()

            for term in matched_terms:
                docs.update(inverted_index[term])

            print("\nDocuments Found:")
            for doc in docs:
                print(doc, ":", documents[doc])
        else:
            print("No matching terms found")

    elif choice == "4":
        print("Thank you")
        break

    else:
        print("Invalid Choice.")

```
